```python
import pandas as pd

# File name
CSV_PATH = "Titanic DataSet.csv"      # Change to your CSV file name

# Read CSV file
df = pd.read_csv(CSV_PATH)

# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())

# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)

# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display dataset information
print("\nDATASET INFORMATION")
df.info()

# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())

# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 9. Passenger Class Distribution
print("\nPASSENGER CLASS DISTRIBUTION")
print(df["Pclass"].value_counts())

# 10. Gender Distribution
print("\nGENDER DISTRIBUTION")
print(df["Sex"].value_counts())

# 11. Gender Percentage
print("\nGENDER PERCENTAGE")
print(df["Sex"].value_counts(normalize=True) * 100)

# 12. Missing Age Values
print("\nMISSING AGE VALUES")
print(df["Age"].isnull().sum())
```
